# PHOENIX spectrum classification: a simple UVB example

> **Notebook role:** Public tutorial / first worked classification example.


This notebook shows a simple PHOENIX-based fitting workflow in Spyctres using a reduced X-SHOOTER UVB spectrum. It uses the current modular package structure: instrument data are read into a canonical `SpectrumSegment`, the PHOENIX templates are handled by `PhoenixLibrary`, the native-grid forward model applies the RV and LSF terms explicitly, and diagnostics report wavelength-frame and grid-boundary assumptions.

This is a compact demonstration, not a final precision analysis. It is designed to show the main moving pieces with a small Balmer-window fit.

## 1. Imports

We import the core Spyctres interfaces, a few plotting helpers, and standard scientific Python tools.


````python
from pathlib import Path
import sys
import warnings

# Let the notebook work both from an installed Spyctres package and from
# a local development checkout. This is only path setup; it does not affect
# the fitting logic below.
repo_root = None
for candidate in [Path.cwd().resolve(), *Path.cwd().resolve().parents]:
    if (candidate / "Spyctres").exists() and (candidate / "setup.py").exists():
        repo_root = candidate
        break
if repo_root is not None and str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

# This will suppress some obsolete warnings
warnings.filterwarnings(
    "ignore",
    message=r"pkg_resources is deprecated as an API.*",
    category=UserWarning,
    module=r"pysynphot.*",
)

import numpy as np
import matplotlib.pyplot as plt

import Spyctres
from Spyctres import format_fit_quality_report, prepare_phoenix_fit_kwargs
from Spyctres.config import load_user_config, get_config_value, resolve_setting
from Spyctres.io import read_spectrum, make_padded_window_segments
from Spyctres.phoenix import PhoenixLibrary
from Spyctres.waveutils import convert_wavelength_medium
from Spyctres.fitting import (
    fit_phoenix_full_spectrum,
    build_effective_fit_mask,
    reconstruct_phoenix_legendre_models_for_segments,
)
from Spyctres.plotting import plot_binned_spectra, plot_full_spectrum_fit
from Spyctres.recipes import pick_grid_range

def pretty_path(path):
    path = Path(path).resolve()
    if repo_root is not None:
        try:
            return str(path.relative_to(repo_root))
        except ValueError:
            pass
    return "<local path>"

print("Using Spyctres from:", pretty_path(Spyctres.__file__))
````

## 2. Locate the PHOENIX templates and the example data

The PHOENIX path is resolved from the user config or the `SPYCTRES_PHOENIX_DIR` environment variable.

The notebook can be run either from the repo root or from the `examples/` directory.


````python
# Read the user's stable local configuration first. The PHOENIX grid is large,
# so Spyctres expects users to keep it outside the repository and point to it
# via SPYCTRES_PHOENIX_DIR or ~/.config/spyctres/config.toml.
config = load_user_config()
phoenix_dir_cfg = get_config_value(config, "paths", "phoenix_dir", default=None)
phoenix_dir = resolve_setting(
    cli_value=None,
    env_var_name="SPYCTRES_PHOENIX_DIR",
    config_value=phoenix_dir_cfg,
    default=None,
)
if phoenix_dir is None:
    raise RuntimeError(
        "No PHOENIX directory found. Set SPYCTRES_PHOENIX_DIR or "
        "[paths].phoenix_dir in ~/.config/spyctres/config.toml."
    )

cwd = Path.cwd().resolve()
if (cwd / "examples" / "data").exists():
    data_dir = cwd / "examples" / "data"
elif (cwd / "data").exists():
    data_dir = cwd / "data"
else:
    raise FileNotFoundError(
        "Could not find the example data directory. Run this notebook from the repo root "
        "or from the examples/ directory."
    )

spectrum_path = data_dir / "TOO_Gaia21ccu_SCI_SLIT_FLUX_MERGE1D_UVB.fits"

print("Spectrum path:", pretty_path(spectrum_path))
print("PHOENIX dir :", "<configured PHOENIX directory>")
````

## 3. Read the UVB spectrum and inspect the metadata

The X-SHOOTER reader stores basic metadata such as the arm, wavelength medium, frame, and an inferred nominal resolving power.


````python
# All instrument-specific FITS details are handled by the reader. The returned
# SpectrumSegment is Spyctres' common internal format: wavelength, flux,
# uncertainty, mask, resolution, and wavelength/frame metadata travel together.
seg0 = read_spectrum(str(spectrum_path), instrument="xshooter")

print("Name                :", seg0.name)
print("Object              :", seg0.meta.get("object"))
print("Instrument          :", seg0.meta.get("instrument"))
print("Arm                 :", seg0.meta.get("arm"))
print("Wave medium         :", seg0.wave_medium)
print("Legacy wave_frame   :", seg0.wave_frame)
print("Observer frame      :", seg0.observer_frame)
print("Stellar rest status :", seg0.stellar_rest_status)
print("Resolution metadata :", None if seg0.resolution is None else seg0.resolution.to_metadata())
print("Barycorr [km/s]     :", seg0.meta.get("barycorr_kms"))
print("Pixels              :", len(seg0.wave))
print("Wave range [Å]      :", float(np.min(seg0.wave)), float(np.max(seg0.wave)))
````

## 4. Quick look at the spectrum

We use a median-binned view here so the overall shape and the main absorption lines are easier to see.


````python
# These broad windows are the parts of the UVB spectrum we will inspect and
# fit in this tutorial. They focus on Balmer-line wings rather than the whole
# spectral energy distribution.
fit_windows = [
    ("Hdelta", 3980.0, 4220.0),
    ("Hgamma", 4220.0, 4480.0),
    ("Hbeta", 4700.0, 5020.0),
]

fig, ax = plot_binned_spectra(
    spectra=[seg0],
    nbins=700,
    title="Gaia21ccu X-SHOOTER UVB quick look",
    highlight_regions=[(w0, w1) for _, w0, w1 in fit_windows],
)
plt.show()
````

## 5. Choose the wavelength windows and first-pass fit defaults

For this example we focus on three broad hydrogen lines in the UVB arm: Hdelta, Hgamma, and Hbeta.

Spyctres can now suggest conservative first-pass PHOENIX fitting defaults from the loaded spectrum metadata and wavelength coverage. This notebook uses that helper, then deliberately overrides the window and atmospheric bounds for this hot-star/Balmer demonstration. The printed reasons and warnings are part of the example: they show which assumptions are automatic and which ones are tutorial choices.

````python
# Ask Spyctres for conservative first-pass defaults, then override the few
# choices that make this hot-star/Balmer-wing tutorial clearer. Expert users
# can make the same overrides in scripts or a future GUI.
fit_kwargs, default_suggestion = prepare_phoenix_fit_kwargs(
    seg0,
    defaults_mode="quicklook",
    science_case="classification",
    p0_overrides=(10000.0, -0.5, 3.0, -5.0),
    lower_bound_overrides=(9000.0, -2.0, 2.0, None),
    upper_bound_overrides=(11600.0, 0.0, 4.5, None),
    window=(3980.0, 5500.0),
    extra_kwargs={
        "mdeg": 2,
        "rv_grid_n": 61,
        "coarse_teff_grid": [9000.0, 9800.0, 10600.0, 11600.0],
        "coarse_feh_grid": [-1.0, -0.5, 0.0],
        "coarse_logg_grid": [2.5, 3.0, 3.5, 4.0, 4.5],
    },
)

print("Suggested/defaulted fit choices:")
for reason in default_suggestion.reasons:
    print(" -", reason)
for warning in default_suggestion.warnings:
    print(" WARNING:", warning)
interpretation = default_suggestion.provenance.get("interpretation", {})
if interpretation:
    print(" interpretation intended use:", interpretation.get("intended_use"))
    print(" interpretation RV role     :", interpretation.get("rv_role"))
    print(" interpretation risk flags  :", interpretation.get("risk_flags"))
print("\nTutorial overrides now in use:")
print(" p0     :", fit_kwargs["p0"])
print(" bounds :", fit_kwargs["bounds"])
print(" regions:", fit_kwargs["regions"])
print(" rv_grid_n:", fit_kwargs["rv_grid_n"])

wmin, wmax = fit_kwargs["regions"][0]
window_pad = 5.0

# First clip to the broad selected range, then split into named line windows.
# The padded support around each line helps the continuum polynomial behave
# smoothly near the edges of each fitted segment.
seg_clip = seg0.window(
    wmin=wmin,
    wmax=wmax,
    clip_left=0.0,
    clip_right=0.0,
    name_suffix="fitwin",
)

segments = make_padded_window_segments(
    seg_clip,
    fit_windows,
    pad=window_pad,
)

for seg_i, (label, _w0, _w1) in zip(segments, fit_windows):
    seg_i.name = label

print("Clipped segment:")
print(" name       :", seg_clip.name)
print(" pixels     :", len(seg_clip.wave))
print(" wave range :", float(np.min(seg_clip.wave)), float(np.max(seg_clip.wave)))

print("\nPer-window segments:")
for s in segments:
    print(
        f" {s.name:>6s} : "
        f"pixels={len(s.wave):4d}  "
        f"support=[{float(np.min(s.wave)):.1f}, {float(np.max(s.wave)):.1f}] Å  "
        f"fit_pixels={int(np.sum(s.mask))}"
    )
````

## 6. Exclude the very line centers

For a simple first fit, it is often useful to exclude a narrow region around the center of each Balmer line and fit mainly the wings.

Here we use a symmetric half-width of 6 Å around each line center.


````python
from Spyctres.recipes import make_balmer_core_exclude_mask

# The Balmer cores can be affected by physics this simple example does not fit
# explicitly. Masking only the narrow cores lets the broad wings drive the
# first-pass atmospheric classification.
# For the core_mask_halfwidth, try 6, 8, 10, 12 Angstrom as a robustness check. 
# Stable parameters across this range are more trustworthy than a single fit.
core_mask_halfwidth = 6.0
exclude_mask = make_balmer_core_exclude_mask(
    core_halfwidth=core_mask_halfwidth,
    wave_medium=segments[0].wave_medium,
)

used_masks_plot = [
    build_effective_fit_mask(seg, exclude_mask=exclude_mask)
    for seg in segments
]

print("Per-window used pixels:")
for seg_i, mask_i in zip(segments, used_masks_plot):
    print(f" {seg_i.name:>6s}: {int(np.sum(mask_i))} / {len(seg_i.wave)}")
print("\nTotal used pixels:", int(sum(np.sum(m) for m in used_masks_plot)))
````

## 7. Define the PHOENIX grid

The fit-default helper gives us broad parameter bounds. For this worked example we convert those bounds into an explicit installed PHOENIX subgrid. This keeps the tutorial fast and makes it obvious which model nodes were available to the interpolator.

For real work, start broad enough that the solution is not pinned to a grid edge, then refine.

````python
# Load the PHOENIX library object once. It knows which templates are installed
# locally and later builds the interpolator/cache needed by the fit.
phoenix_lib = PhoenixLibrary(phoenix_dir, verbose=True)

teff_avail, feh_avail, logg_avail = phoenix_lib.available_axes()
print("Installed Teff range :", float(np.min(teff_avail)), float(np.max(teff_avail)))
print("Installed [Fe/H] range:", float(np.min(feh_avail)), float(np.max(feh_avail)))
print("Installed logg range :", float(np.min(logg_avail)), float(np.max(logg_avail)))

bounds = fit_kwargs["bounds"]
# Restrict the template cube to the requested bounds. This keeps the example
# fast and makes the fitted model domain explicit.
teff_grid_req = pick_grid_range(teff_avail, bounds[0][0], bounds[1][0])
feh_grid_req = pick_grid_range(feh_avail, bounds[0][1], bounds[1][1])
logg_grid_req = pick_grid_range(logg_avail, bounds[0][2], bounds[1][2])

teff_grid_fit, feh_grid_fit, logg_grid_fit = phoenix_lib.complete_subgrid(
    teff_grid_req,
    feh_grid_req,
    logg_grid_req,
)

print("\nTeff grid :", teff_grid_fit)
print("FeH grid  :", feh_grid_fit)
print("logg grid :", logg_grid_fit)
````

## 8. Run the fit

The fit solves for:
- effective temperature `Teff`
- metallicity `[Fe/H]`
- surface gravity `logg`
- radial velocity `RV`

It also solves a low-order multiplicative continuum correction for each segment during the fit.


````python
# Use the reader's nominal resolving power when available. This gives the
# forward model a simple Gaussian instrumental broadening, but it is not a
# detailed slit/seeing/time-dependent LSF model.
R = seg0.meta.get("resolution_R", None)
# For simplicity, this example does not apply an additional barycentric correction during
# the fit, even though the reader reports one in the metadata.
# We set rv_bary_kms = 0.0 because it keeps the example focused on spectral classification.
# The fitted RV should therefore be read as an internal alignment parameter,
# not a final barycentric radial velocity.
rv_bary_kms = 0.0
# For a barycentric RV-oriented run, use:
# rv_bary_kms = seg0.meta.get("barycorr_kms", 0.0)
p0 = fit_kwargs["p0"]

print("R used            :", R)
print("rv_bary_kms used  :", rv_bary_kms)
print("Initial guess p0  :", p0)
print("Bounds            :", fit_kwargs["bounds"])


def notebook_progress(message):
    # Long PHOENIX cache/interpolation steps can otherwise look frozen in a
    # notebook. The callback prints progress messages from the fitter.
    print(f"[Spyctres] {message}", flush=True)


# This is the explicit low-level fit call used for teaching. The public API
# shown at the end wraps the same ideas with fewer knobs.
result = fit_phoenix_full_spectrum(
    segments=segments,
    phoenix_lib=phoenix_lib,
    p0=p0,
    bounds=fit_kwargs["bounds"],
    exclude_mask=exclude_mask,
    mdeg=fit_kwargs["mdeg"],
    rv_bary_kms=rv_bary_kms,
    R=R,
    forward_model=fit_kwargs["forward_model"],
    teff_grid=teff_grid_fit,
    feh_grid=feh_grid_fit,
    logg_grid=logg_grid_fit,
    cache_path="/tmp/spyctres_example_xshooter_cache.npz",
    physical_init=fit_kwargs.get("physical_init"),
    coarse_teff_grid=fit_kwargs.get("coarse_teff_grid"),
    coarse_feh_grid=fit_kwargs.get("coarse_feh_grid"),
    coarse_logg_grid=fit_kwargs.get("coarse_logg_grid"),
    coarse_decimate=fit_kwargs.get("coarse_decimate", 12),
    multistart=fit_kwargs.get("multistart", 1),
    rv_init=fit_kwargs.get("rv_init"),
    rv_grid_n=fit_kwargs["rv_grid_n"],
    verbose=1,
    max_nfev=300,
    progress_callback=notebook_progress,
)

print("\nFit result summary:")
for key in ["teff", "feh", "logg", "rv_kms", "chi2", "dof", "chi2_red", "success", "message"]:
    print(f"{key:>8s} : {result[key]}")
print()
print(format_fit_quality_report(result))
````

## 9. Reconstruct and inspect the fitted model

This is the main diagnostic step. The plots show the data, the fitted model, and which pixels were actually used in the fit.


````python
# Reconstruct the best-fit model on the observed wavelength grid. The fitter
# solves continuum coefficients internally, so this step recreates the model
# that should be compared with the data in each plotted segment.
model_list, coeffs_list, used_masks, excluded_masks = reconstruct_phoenix_legendre_models_for_segments(
    segments=segments,
    phoenix_lib=phoenix_lib,
    fit_result=result,
    exclude_mask=exclude_mask,
    rv_bary_kms=rv_bary_kms,
    R=R,
    mdeg=fit_kwargs["mdeg"],
    forward_model=fit_kwargs["forward_model"],
)

for seg_i, model_i, used_i, excl_i in zip(segments, model_list, used_masks, excluded_masks):
    title = (
        f"{seg_i.name}   "
        f"Teff={result['teff']:.0f} K   "
        f"[Fe/H]={result['feh']:.2f}   "
        f"logg={result['logg']:.2f}   "
        f"RV={result['rv_kms']:.1f} km/s   "
        f"chi2_red={result['chi2_red']:.2f}"
    )
    fig, axes = plot_full_spectrum_fit(
        wave=seg_i.wave,
        flux=seg_i.flux,
        err=seg_i.err,
        model=model_i,
        used_mask=used_i,
        excluded_mask=excl_i,
        title=title,
        line_groups=["balmer"],
    )

    ax_top = axes[0]
    y_min = min(np.nanmin(seg_i.flux), np.nanmin(model_i))
    y_max = max(np.nanmax(seg_i.flux), np.nanmax(model_i))
    y_pad = 0.05 * (y_max - y_min)
    ax_top.set_ylim(y_min - y_pad, y_max + y_pad)
    plt.show()
````

The shaded region is excluded from the fit. A mismatch inside the shaded core should not be interpreted as a failed fit; the relevant comparison is mainly in the unshaded wings.

````python
# Assess fit quality before trusting the classification. This report is more
# important than any single fitted parameter: it tells you about masks,
# grid-edge behavior, frame assumptions, residual structure, and other caveats.
print(format_fit_quality_report(result))
diagnostics = result.get("diagnostics", {})
grid_edge_flags = diagnostics.get("grid_edge_flags", {})
print("grid-edge flags:", grid_edge_flags if grid_edge_flags else "not reported")
print("velocity convention:", diagnostics.get("velocity_convention", result.get("rv_sign_convention", "not reported")))
print("line-window residuals should be inspected visually before adopting the classification.")
````

## 10. Summarize the result

This is a compact text summary that is easy to copy into notes.


````python
fit_summary = {
    "Teff [K]": result["teff"],
    "[Fe/H]": result["feh"],
    "logg": result["logg"],
    "RV fit [km/s]": result.get("rv_kms_fit", result["rv_kms"]),
    "Input barycentric term [km/s]": result.get("rv_bary_kms_input", result.get("rv_bary_kms")),
    "Total model shift [km/s]": result.get("total_model_shift_kms"),
    "Reduced chi^2": result["chi2_red"],
}

print("Generic PHOENIX fit summary:")
for key, value in fit_summary.items():
    if isinstance(value, (int, float, np.floating)):
        print(f" {key:>30s} : {value:.3f}")
    else:
        print(f" {key:>30s} : {value}")

print("\nInterpretation:")
print(
    f"The spectrum is matched by a model with Teff {result['teff']:.0f} K, "
    f"[Fe/H] {result['feh']:.2f}, logg {result['logg']:.2f}, "
    f"and fitted stellar RV {result['rv_kms']:.1f} km/s."
)
print(
    "This should be read as a first model-based classification from a simple workflow, "
    "not as a final precision measurement. Positive RV redshifts the stellar model; "
    "the barycentric term is kept explicit in the result provenance."
)
````

## Interpretation summary

This public tutorial demonstrates a compact PHOENIX-backed classification of one reduced X-SHOOTER UVB spectrum. It is intended as a first worked example, not a final precision analysis. The fitted `rv_kms` is an internal alignment parameter here because the notebook deliberately sets `rv_bary_kms = 0.0`; positive stellar RV redshifts the model spectrum. Trust the broad classification and the diagnostic behavior first, and inspect the quality flags, grid-edge flags, residuals, and wavelength/frame metadata before treating the parameters as physically final.


## 11. What to try next

A few natural follow-up tests are:
- revise the fitted wavelength windows
- try a different Balmer-core mask width
- test different PHOENIX subgrids
- inspect each line separately
- compare different continuum treatments
- revisit the assumed resolving power or wavelength handling if the residuals suggest a mismatch

For many real spectra, fitting is iterative: inspect the residuals, adjust the modeling choices, and fit again.

## The five-line API path

The user-facing mental model is much shorter than the teaching workflow above:

```python
from Spyctres import read_spectrum, fit_stellar_spectrum, plot_fit_referee

spec = read_spectrum("my_spectrum.fits", instrument="xshooter")
result = fit_stellar_spectrum(spec, model="phoenix")
plot_fit_referee(result)
```

Use the longer notebook cells when you want to understand or customize the wavelength windows, masks, PHOENIX subgrid, RV handling, and quality diagnostics.
